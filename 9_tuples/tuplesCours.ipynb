## Définition


```python
t1 = (1, 4, "bonjour", 4.6)
t2 = ()  # un tuple vide
t3 = ("a",)  # un tuple à 1 élément (la virgule est obligatoire)
```

```python
t1[2] = "hello"
```

## Opérations sur les tuples


```python
print("bonjour" in t1)
print(len(t1))
print(t1 == t2)
print(t1[1])
print(t1 + t3)
print(t3 * 4)
```

## Parcours de tuple


```python
t4 = ("informatique", "python", "réseau", "javascript")
for mot in t4:
    print(mot)
```

## Quelques utilisations des tuples


```python
def milieu(p1: tuple, p2: tuple) -> tuple:
    x1, y1 = p1
    x2, y2 = p2
    xm = (x1 + x2) / 2
    ym = (y1 + y2) / 2
    return (xm, ym)


a, b = milieu((0, 1), (1, 0))
print(f"Les coordonnées du milieu est : ({a}, {b})")
```
