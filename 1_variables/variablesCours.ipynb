## Types

1. Types **primitifs** :
   - entiers : `int`
   - flottants : `float`
   - booléens : `bool`
   - chaînes de caractères : `str`
2. Types **composés** : ...


```python
print(2022, type(2022))
print(12.5, type(12.25))
print(True, type(True))
print("bonjour", type("bonjour"))
```

## Opérateurs


### Opérateurs arithmétiques

| Opérateurs | Signification    |
| ---------- | ---------------- |
| +          | addition         |
| -          | soustraction     |
| \*         | multiplication   |
| /          | division         |
| //         | division entière |
| %          | modulo           |
| \*\*       | puissance        |


```python
print(3 + 5, type(3 + 5))  # 3 et 5 sont des entiers donc le résultat est un entier
print(3.0 + 5, type(3.0 + 5))  # 3. est un flottant donc le résultat est un flottant
print(10 / 3, type(10 / 3))  # le résultat d'une division est toujours un flottant
print(10 // 3, type(10 // 3))  # le quotient de la division euclidienne : 10 = 3 * 3 + 1
print(10 % 3, type(10 % 3))  # le reste de la division euclidienne : 10 = 3 * 3 + 1
```

### Opérateurs de chaînes de caractères

| Opérateurs | Signification |
| ---------- | ------------- |
| +          | concaténation |
| \*         | duplication   |


```python
print("bonjour " + "tout le monde")
print("hello" * 5)
```

### Opérateurs de comparaison

| Opérateurs | Signification     |
| ---------- | ----------------- |
| <          | inférieur         |
| <=         | inférieur ou égal |
| >          | supérieur         |
| >=         | supérieur ou égal |
| ==         | égal              |
| !=         | différent         |


```python
print(3 < 9.5)  # il est possible de comparer un entier et un flottant
print(
    "bonjour" < "salut"
)  # les chaines de caractères sont comparées avec l'ordre alphabétique
print(
    3 < "a"
)  # erreur : il n'est pas possible de comparer un entier et une chaîne de caractères
```

### Opérateurs booléens

| Opérateurs | Signification |
| ---------- | ------------- |
| not        | non           |
| and        | et            |
| or         | ou            |


```python
print((3 == 3) or (9 > 24))
print((9 > 24) and (3 == 3))
```

## Conversions de types

Fonctions de conversion de types:

- `int`: conversion en entier
- `float`: conversion en flottant
- `str`: conversion en chaîne de caractères.


```python
x = int(1.999)  # la conversion d'un entier en flottant ne garde que la partie entière
print(x)
s = str(3.14)
t = s + s  # le + fait une concaténation
print(t)
a = int("3")  # la chaîne '3' a été convertie en entier 3
print(a, type(a))
print(
    int("hello")
)  # si la chaîne de caractères ne contient pas d'entier, il n'est pas possible de la convertir
```

## Entrées/sorties

### Sorties

On utilise la fonction `print`.


```python
print(3.5)
print("Bonjour")

nom = "John"
age = 30
print(nom, "a", age, "ans")

print(f"{nom} a {age} ans")  # f-string

print("Bonjour", end=" ")
print("Jean-Paul")

print(3, 5, 7, sep="*")
```

## Entrées

On utilise la fonction `input`.


```python
nom = input("Entrez votre nom: ")
print("Bonjour", nom)
```
