## Déclaration d'une fonction

```python
def hello():
    print("Hello world!")


hello()
```

## Paramètres d'une fonction


```python
def hello(name):
    print(f"Hello {name} !")


hello("Alice")
hello("Bob")
```

## Renvoi d'une valeur


```python
def max1(a: int, b: int) -> int:
    if a > b:
        return a
    else:
        return b


print(max1(3, 7))
print(max1(9, 4))
```

```python
print(max1(max1(4, 9), 6))
```

```python
def max2(a: int, b: int):
    if a > b:
        print(a)
    else:
        print(b)


max2(3, 7)
max2(9, 4)
```

```python
print(max2(max2(4, 9), 6))
```
